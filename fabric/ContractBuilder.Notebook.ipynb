# Generate mapping contracts in Microsoft Fabric

Attach a default Lakehouse and the environment containing the `fabric-data-platform-framework` wheel. Place the workbook under the Lakehouse `Files` area. The build writes metadata, draft contracts, and SQL under `Files/contract-build/`; it does not deploy or execute SQL.

```python
workbook_path = "s2t.xlsx"
output_dir = "contract-build"
discover_source_schema = False
```

```python
from pathlib import Path

from framework.fabric import run_fabric_build

generated_paths = run_fabric_build(
    Path("/lakehouse/default/Files"),
    workbook_path=workbook_path,
    output_dir=output_dir,
    discover_source_schema=discover_source_schema,
)

print(f"Generated {len(generated_paths)} artifacts under Files/{output_dir}")
```
